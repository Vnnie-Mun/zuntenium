### Step 2: Import & Verify Environment

```python
import zuntenium
from zuntenium.core import Circuit
from zuntenium.simulators import StatevectorSimulator, plot_histogram

print(f"✓ Successfully loaded Zuntenium v{zuntenium.__version__}")
```

### Step 3: Create a Bell State (|Φ⁺⟩ = (|00⟩ + |11⟩) / √2)

```python
# Allocate 2-qubit circuit
circ = Circuit(2, name="bell_state")

# Apply Hadamard and CNOT gates
circ.h(0)
circ.cx(0, 1)

# Display the circuit directly in Colab
circ
```

### Step 4: Simulate with StatevectorSimulator

```python
sim = StatevectorSimulator()
result = sim.run(circ, shots=1024)

print("Simulation Counts:", result.counts)
```

### Step 5: Visualize Measurement Distribution

```python
# Plot bar chart directly in the notebook
result.plot_histogram()
```

### Step 6: 3-Qubit Quantum Teleportation Protocol

```python
# Q0 = State to teleport, Q1 = Alice EPR, Q2 = Bob EPR
teleport = Circuit(3, name="teleportation")

# 1. State preparation on Q0
teleport.h(0)

# 2. Shared Bell pair between Q1 & Q2
teleport.h(1).cx(1, 2)

# 3. Alice Bell measurement
teleport.cx(0, 1).h(0)

# Render circuit
teleport
```

```python
# Execute teleportation simulation
teleport_res = sim.run(teleport, shots=1024)
print("Teleportation Counts:", teleport_res.counts)
teleport_res.plot_histogram()
```
